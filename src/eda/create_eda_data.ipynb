```python
import pandas as pd
from constants import CATEGORICAL_COLS
```

```python
cleaned_data_path = "../../data/cleaned_t2dm_a2h.tsv"
data = pd.read_csv(cleaned_data_path, sep='\t')
```

```python
data['group_id'] = data['pmcid'] + "_" + data['NCT Number']
data = data.set_index('group_id')
```

```python
data['result'] = data['clinical_change_of_hba1c_value'] - data['preclinical_change_of_hba1c_value']
```

```python
data.columns
```

```python
drop_cols = ['pmcid', 'link', 'preclinical_year',
             'NCT Number', 'preclinical_treatment_name', 'gender',
             'clinical_change_of_hba1c_value',
             'preclinical_change_of_hba1c_value',
              'result_translation_success']

data = data.drop(columns=drop_cols)
```

```python
data.rename(columns={'clinical_treatment_name': 'intervention',
                     "animal_strain": "preclinical_animal_strain",
                     "animal_species": "preclinical_animal_species",
                     "animal_sex": "preclinical_animal_sex",
                     "disease_induction_method": "preclinical_disease_model",
                     'animal_weight_value(g)': 'preclinical_animal_weight_value(g)',
                     'animal_age(days)': "preclinical_animal_age(days)",
                     "age_groups": "clinical_age_groups",
                     "phases": "clinical_phases",
                     "sample_size": "clinical_sample_size"}, inplace=True)
```

```python
for cat_col in CATEGORICAL_COLS:
    data[cat_col] = data[cat_col].fillna('Unknown')
```

```python
data.to_csv("cleaned_t2dm_a2h_eda.tsv", sep='\t')
```
